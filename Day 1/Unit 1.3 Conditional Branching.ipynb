## Boolean Values 

In everyday life, we make true/false decisions all the time — ‘Is the light on?’, ‘Did I finish my homework?’, ‘Is it raining?’ 
These are all yes/no, true/false questions — that’s what computers do too. They use Booleans to make decisions.

Booleans are data types that have only two possible values: ```True``` and ```False```, or ```1``` and ```0```. They represent the two truth values of Boolean logic and algebra (named after the English mathematician, philosopher and logician George Boole, who worked on this logic in the 19th century).

In Python we can assign the values ```True``` and ```False``` to variables just like strings and numbers, and also use and for example print them as usual:

````python
a = True
b = False
print("a is", a, "and b is", b) 
````

What to remember: 
- The words True and False must have the first letter capitalized.
- They are their own special data type.
- They are not the same as the strings "True" or "False".

````python
print(type(True))     # bool
print(type("True"))   # str
````

## Boolean Expressions



Boolean algebra has three basic operations: ```not```, ```and``` and ```or```(evaluation in this order of precedence). The following Python program demonstrates how they work:

````python
a = True
b = False
print("a is", a, "and b is", b)
print("not", a, "is", not a, "and not", b, "is", not b)
print(a, "or", b, "is", a or b)
print(a, "and", b, "is", a and b)
````

The three basic operations function as follows:
- The ```not``` operation turns a Boolean value into the other. 
- The result of an ```or``` operation with two Boolean values is ```True``` if at least one of them is ```True```. 
- The result of an ```and``` operation is ```True``` if both input values are ```True```. 

Typically the functional values of Boolean expressions for all possible combinations of values of their functional arguments are surveyed in truth tables, like shown below:


|```a```|```b```|```not a```|```not b```|```a or b```|```a and b```|
| :-: | :-: | :-: | :-: | :-: | :-: |
|```True```|```True```|```False```|```False```|```True```|```True```|
|```True```|```False```|```False```|```True```|```True```|```False```|
|```False```|```True```|```True```|```False```|```True```|```False```|
|```False```|```False```|```True```|```True```|```False```|```False```|

In real programs, we rarely just write True or False directly. Usually, a question or condition produces a Boolean.

Usually Boolean values originate from comparisons or other tests. In Python the basic comparison operators returning ```True``` or ```False``` are:

	<	 (less than)
	>	 (greater than)
	<=	(less than or equal to)
	>=	(greater than or equal to)
	==	(equal to)
	!=	(not equal to)
    
Equality and non-equality can be checked for objects of all data types (also those that we have not yet discussed), while the less/greater than checks only work for data types for which an ordering relation is defined. This is for example the case for numbers (integers and floats) and strings, so they can easily be applied there:

````python
a = 5.2
b = 3
print(a, "<", b, "is", a < b)
print(a, "<=", b, "is", a <= b)
print(a, ">", b, "is", a > b)
print(a, ">=", b, "is", a >= b)
print(a, "==", b, "is", a == b)
print(a, "!=", b, "is", a != b)
````

Let's see some string comparisons:  

````python
a = "hello"
b = "world"

print(a, "==", b, "is", a == b)
print(a, "!=", b, "is", a != b)
````

Note that comparisons can be chained arbitrarily, that is, and expression like ```0 < x <= 10``` is perfectly valid, but this is often not easy to read and understand. In most cases it is thus better coding style to use only pair-wise comparisons and chain them together with the Boolean `and` operator:

````python
x = int(input("Enter an integer number: "))
is_in_range = 0 < x and x <= 10 #same as is_in_range = 0 < x <= 10
print("Is", x, "in range from 0 to 10?", is_in_range) 
````

This example also shows how comparisons that result in Boolean values and Boolean operations can be combined into more complex Boolean expressions. Here another, maybe more meaningful example:

````python
weekday = input("Which day of the week is it? ")
is_lectureday = (weekday == "Wednesday" or weekday == "Friday")
print("Lecture day?", is_lectureday)
````


Python does not always check every part of a Boolean expression. It checks from left to right and stops as soon as it already knows the final answer. This is called "lazy evaluation".

For example, imagine we have an or condition that checks different days:
```day == "Monday" or day == "Tuesday" or day == "Wednesday"```

If the day is ```Tuesday```, Python checks the first condition, then the second, and then stops. Once it finds that ```day == "Tuesday"``` is ```True```, it stops checking. Python already knows that the whole or expression is ```True```, so it does not need to continue.

Taking into account this lazy evaluation behavior can help to write more efficient evaluations, for example by checking those things first that can already determine the result of the complete expression, or by performing "expensive" calls (e.g. to external databases) only when it is really needed for determining the outcome. 


## The `if` statement

Now let’s see how we can use Boolean expressions for implementing conditional branching in Python programs. This will allow us to not only print out the bare Boolean values as we did above, but to actually react on whether something is true or false. What we need is the ```if``` statement. The basic structure is the following:

```
if <some condition>:
    <do something>
```

> An if statement checks if a condition is True. If it is, it runs a block of code. If it’s False, it skips it.

Here is a simple example:

````python
name = input("What is your name? ")
if name == "Jane":
    print("Hello Jane, nice to see you again!") 
````

In the first line the program reads an input (name) from the user. In the next line, it checks if the name is “Jane”, and only if that is ```True``` the following line is executed. If it is ```False``` (the user did not enter Jane), nothing else happens.

If we want something to happen also for the case the checked condition is ```False```, we can add an ```else``` statement. The structure becomes like this:

```
if <some condition>:
    <do something>
else:
    <do something else>
```

For example:

````python
name = input("What is your name? ")
if name == "Jane":
    print("Hello Jane, nice to see you again!")
else:
    print("Hello", name, "nice to meet you!") 
````

If we want to distinguish even more than two cases, we can furthermore add ```elif``` (Python’s abbreviation for “else if”) statements. For example:

````python
name = input("What is your name? ")
if name == "Jane":
    print("Hello Jane, nice to see you again!")
elif name == "Steven":
    print("Hello Steven, nice to see you again!")
elif name == "Kim":
    print("Hello Kim, nice to see you again!")
else:
    print("Hello", name, "nice to meet you!")
````

With the if statement we can also rewrite the Boolean expression examples from above to produce more complete
outputs:

````python
x = int(input("Enter an integer number: "))
if 0 < x and x <= 10:
    print(x, "is in range from 0 to 10.")
else:
    print(x, "is not in range from 0 to 10")
````

````python
weekday = input("Which day of the week is it? ")
if (weekday == "Wednesday" or weekday == "Friday"):
    print("It is lecture day!")
elif (weekday == "Saturday" or weekday == "Sunday"):
    print("It is weekend!")
else:
    print("It is study day!") 
````

So, the complete structure for conditional branching with in Python can be described as follows (where ```elif``` and ```else``` are optional and there can be more than one ```elif```):

```
if <some condition>:
    <do something>
elif <some condition>:
    <do something else>
else:
    <do something else>
```

Two things are important to note here:

**1. There can be more than one line of code after the ```if/elif/else``` statements** that is executed for the respective case. These lines of code together are then called a block. Blocks can contain further ```if/elif/else``` statements.


**2. Indentation is important.** The ```if/elif/else``` blocks are indented by exactly four spaces (in most Python editors the TAB key will insert them automatically) to denote that they belong to the block for the respective case. The next code after the if-statement is simply not indented. (We will see later that Python uses this “meaningful whitespace” also for other constructs. Many other languages use additional parentheses for the same purpose.)

Here is an example to illustrate the above. We show an example in two versions; without and with nested if

````python
# Ask for the user's age and if they have a ticket without nested if
age = int(input("How old are you? "))
has_ticket = input("Do you have a ticket? (yes/no): ")

if age >= 18 and has_ticket == "yes":
    print("You are old enough to enter.")
    print("Welcome to the event!")
elif age >= 18 and has_ticket != "yes":
    print("You are old enough to enter.")
    print("Sorry, you need a ticket to enter.")
else:
    print("You are too young to enter the event.")
````

````python
# Ask for the user's age and if they have a ticket with nested if

age = int(input("How old are you? "))
has_ticket = input("Do you have a ticket? (yes/no): ")

if age >= 18:
    print("You are old enough to enter.")

    # Nested if: check if they also have a ticket
    if has_ticket == "yes":
        print("Welcome to the event!")
    else:
        print("Sorry, you need a ticket to enter.")
else:
    print("You are too young to enter the event.")
````

## Takeaway: What You Learned 

👏 Great work! Here’s what you should remember from this lesson:

- **Booleans** are a special data type with only two possible values: **`True`** or **`False`**.
- Booleans come from **questions** or **comparisons**, for example, `5 > 3` will give `True`.
- **Comparison operators** like `==`, `!=`, `<`, `>`, `<=`, and `>=` help you check conditions.
- **Logical operators** like `and`, `or`, and `not` let you combine or reverse conditions.
- **`if` statements** run a block of code *only* if a condition is `True`.
- **`elif`** (else if) lets you check more than one condition in order.
- **`else`** runs a block of code when none of the previous conditions are `True`.
- **Indentation** matters — it defines which lines belong to which `if`, `elif`, or `else` block.
- You can **nest `if` statements** inside each other or combine conditions with `and`/`or` to control your program’s logic.
